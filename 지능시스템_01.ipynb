```python
# PyTorch 라이브러리 불러오기
# → GPU(CUDA)를 사용할 수 있는지 확인하기 위해 사용
import torch

# Hugging Face Transformers의 pipeline 함수 불러오기
# → 복잡한 모델 처리 과정을 간단하게 사용할 수 있도록 해주는 함수
from transformers import pipeline


# 감성 분석(Sentiment Analysis) 파이프라인 생성
classifier = pipeline(

    # 수행할 NLP 작업 지정
    # sentiment-analysis: 문장의 감성이 긍정인지 부정인지 분류
    task="sentiment-analysis",

    # 사용할 사전학습 모델 지정
    # DistilBERT를 SST-2 감성 분석 데이터셋으로 파인튜닝한 모델
    model="distilbert-base-uncased-finetuned-sst-2-english",

    # 실행 장치 지정
    # GPU(CUDA)가 있으면 첫 번째 GPU(device=0)를 사용
    # GPU가 없으면 CPU(device=-1)를 사용
    device=0 if torch.cuda.is_available() else -1
)
```

```python
# 감성 분석할 영어 문장 2개
texts = [
    "This lecture is absolutely fascinating!",  # 긍정적인 문장
    "I couldn't understand anything at all."    # 부정적인 문장
]

# 여러 문장을 한 번에 감성 분석
results = classifier(texts)

# 각 문장과 분석 결과를 하나씩 꺼내서 출력
for text, result in zip(texts, results):

    # label : 감성 분석 결과 (POSITIVE 또는 NEGATIVE)
    # score : 모델이 해당 결과를 얼마나 확신하는지 나타내는 점수
    # :.4f  : 점수를 소수점 4자리까지 출력
    print(f"{result['label']:8s} {result['score']:.4f} <- {text}")
```

```python
# Hugging Face에서 자동으로 토크나이저를 불러오는 클래스
from transformers import AutoTokenizer

# 여러 언어를 처리할 수 있는 다국어 BERT 토크나이저 불러오기
tok = AutoTokenizer.from_pretrained(
    "google-bert/bert-base-multilingual-cased"
)

# 영어와 한국어 예제 문장 준비
samples = {
    "영어": "Natural language processing is fascinating.",
    "한국어": "자연어처리는 정말 흥미롭습니다.",
    "중국어": "自然语言处理真的很有趣。"
}

# 영어와 한국어, 중국어 문장을 하나씩 처리
for lang, text in samples.items():

    # 문장을 BERT가 처리할 수 있는 작은 단위인 토큰으로 나누기
    tokens = tok.tokenize(text)

    # 각 토큰을 모델이 사용하는 숫자 ID로 변환
    ids = tok.convert_tokens_to_ids(tokens)

    # 사용한 언어 출력
    print(f"[{lang}]")

    # 원래 문장 출력
    print("원문:", text)

    # 토큰화된 결과 출력
    print("토큰:", tokens)

    # 문장이 몇 개의 토큰으로 나누어졌는지 출력
    print("토큰 수:", len(tokens))

    # 각 토큰에 해당하는 숫자 ID 출력
    print("ID:", ids)

    # 영어와 한국어 결과 사이에 빈 줄 추가
    print()
```

```python
# 하나의 문장을 토큰화하고 간단한 통계 정보를 계산하는 함수
def token_stats(tokenizer, text):

    # 문장을 토큰으로 나누기
    tokens = tokenizer.tokenize(text)

    # 계산한 결과를 딕셔너리 형태로 반환
    return {
        # 원문에 포함된 전체 문자 수
        "chars": len(text),

        # 토큰화된 전체 토큰 수
        "tokens": len(tokens),

        # 토큰 하나가 평균적으로 몇 개의 문자로 이루어졌는지 계산
        "chars_per_token": len(text) / max(len(tokens), 1)
    }


# 앞에서 만든 영어와 한국어, 중국어 예제 문장을 하나씩 처리
for lang, text in samples.items():

    # 각 언어별 토큰 통계 결과 출력
    print(lang, token_stats(tok, text))
```
