```python
# PyTorch 라이브러리 불러오기
# → GPU(CUDA)를 사용할 수 있는지 확인하기 위해 사용
import torch

# Hugging Face Transformers의 pipeline 함수 불러오기
# → 복잡한 모델 처리 과정을 간단하게 사용할 수 있도록 해주는 함수
from transformers import pipeline


# 감성 분석(Sentiment Analysis) 파이프라인 생성
classifier = pipeline(

    # 수행할 NLP 작업 지정
    # sentiment-analysis: 문장의 감성이 긍정인지 부정인지 분류
    task="sentiment-analysis",

    # 사용할 사전학습 모델 지정
    # DistilBERT를 SST-2 감성 분석 데이터셋으로 파인튜닝한 모델
    model="distilbert-base-uncased-finetuned-sst-2-english",

    # 실행 장치 지정
    # GPU(CUDA)가 있으면 첫 번째 GPU(device=0)를 사용
    # GPU가 없으면 CPU(device=-1)를 사용
    device=0 if torch.cuda.is_available() else -1
)
```

```python
# 감성 분석할 영어 문장 2개
texts = [
    "This lecture is absolutely fascinating!",  # 긍정적인 문장
    "I couldn't understand anything at all."    # 부정적인 문장
]

# 여러 문장을 한 번에 감성 분석
results = classifier(texts)

# 각 문장과 분석 결과를 하나씩 꺼내서 출력
for text, result in zip(texts, results):

    # label : 감성 분석 결과 (POSITIVE 또는 NEGATIVE)
    # score : 모델이 해당 결과를 얼마나 확신하는지 나타내는 점수
    # :.4f  : 점수를 소수점 4자리까지 출력
    print(f"{result['label']:8s} {result['score']:.4f} <- {text}")
```
